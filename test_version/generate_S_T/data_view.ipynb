```python
transformed_data_path='D:\pycharm\Py_Projects/battery_SOH_predict/foundation_model/transformed_data'
dataset_list=os.listdir(transformed_data_path)
#print(dataset_list)
new_length=256
config=Config()
for i,dataset in enumerate(dataset_list):
    down_sampled_data_path=os.path.join(transformed_data_path,dataset,f'downsampled_data_{new_length}')
    print(dataset)
    for package in os.listdir(down_sampled_data_path):
        #print(package)
        pkl_path = os.path.join(down_sampled_data_path, package, f'{new_length}_{config.All_Dataset[i]}_all_battery_id_data.pkl')
        soh_path = os.path.join(down_sampled_data_path, package, f'{new_length}_{config.All_Dataset[i]}_all_battery_id_soh.pkl')
        with open(soh_path, 'rb') as file:
            soh_data = pickle.load(file)
        plt.figure(figsize=(10, 5))
        for battery in soh_data.keys():
            plt.plot(soh_data[battery], label=battery)
        plt.xlabel('Cycle')
        plt.ylabel('SOH')
        plt.title(f'{dataset}_{package}')
        plt.legend()
        plt.show()
```
